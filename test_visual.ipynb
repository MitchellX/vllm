```python
import matplotlib.pyplot as plt
import numpy as np

# x-axis labels (prefix cache sizes)
x_labels = ['8k', '16k', '32k', ]
x = np.arange(len(x_labels))

# Example TTFT values in milliseconds
# Replace these with your actual measurements
vllm = np.array([1770, 3312, 7200, ])
dattn_kv_transfer = np.array([500, 1100, 2400])
dattn_execution = np.array([1000, 1200, 1750])

# Bar width
width = 0.35

fig, ax = plt.subplots(figsize=(8, 4))

# vLLM bars (left side)
ax.bar(x - width/2, vllm, width, label='vLLM', color='#1f77b4')

# dAttn bars (right side with stacked segments)
ax.bar(x + width/2, dattn_kv_transfer, width, label='dAttn KV Transfer', color='#ff7f0e')
ax.bar(x + width/2, dattn_execution, width, bottom=dattn_kv_transfer,
       label='dAttn Execution', color='#2ca02c')

# Scaling annotations (arrows and text)
scale_factors = ['1.13×', '1.43×', '1.73×']
scale_positions = [(0.1, 1700), (1.1, 2500), (2.1, 4500), ]
for i, (xi, y) in enumerate(scale_positions):
    ax.annotate(scale_factors[i],
                xy=(xi, y),
                xytext=(xi, y + 100),
                textcoords="data",
                ha='center',
                va='bottom',
                fontsize=12,
                color='salmon',
                arrowprops=dict(arrowstyle='->', color='salmon'))

# Axis and legend formatting
ax.set_ylabel('TTFT Time (ms)', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(x_labels)
ax.set_ylim(0, 7500)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.6)

# set title
ax.set_title('opt-6.7B & PCIE 3.0', fontsize=14)

plt.tight_layout()
plt.show()
```
